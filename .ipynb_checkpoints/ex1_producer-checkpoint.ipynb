```python
import json
from kafka import KafkaProducer

BOOTSTRAP = "kafka:9092"
TOPIC = "weather_stream"

producer = KafkaProducer(
    bootstrap_servers=BOOTSTRAP,
    value_serializer=lambda v: json.dumps(v).encode("utf-8"),  # dict python -> bytes
)

future = producer.send(TOPIC, {"msg": "Hello Kafka"})
meta = future.get(timeout=10)  # on attend la confirmation du broker
print(f"envoyé -> topic={meta.topic} partition={meta.partition} offset={meta.offset}")

producer.flush()
producer.close()
```
